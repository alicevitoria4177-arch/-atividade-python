**Questão 4: Maior sequência de crescimento**

```
nums = [int(input(f"N{i+1}: ")) for i in range(10)]
maior = atual = 1
for i in range(1, 10):
    atual = atual + 1 if nums[i] > nums[i - 1] else 1
    if atual > maior:
        maior = atual
print(nums, "\nMaior sequência crescente:", maior)
```

**Questão 5: Pontos no plano**

```
pts = [(int(input("X: ")), int(input("Y: "))) for _ in range(6)]
q = [0, 0, 0, 0, 0]
m_dist, m_p = -1, None

for x, y in pts:
    d = x**2 + y**2
    if d > m_dist:
        m_dist, m_p = d, (x, y)
    if x > 0 and y > 0: q[0] += 1
    elif x < 0 and y > 0: q[1] += 1
    elif x < 0 and y < 0: q[2] += 1
    elif x > 0 and y < 0: q[3] += 1
    else: q[4] += 1

print(pts, f"\nQuadrantes 1-4/Eixos: {q}\nMais distante: {m_p}")
```

**Questão 6: Frequência e palavra mais repetida**

```

freq = {}
for _ in range(10):
    p = input("Palavra: ")
    freq[p] = freq.get(p, 0) + 1

m_freq = 0
for k in freq:
    if freq[k] > m_freq: m_freq = freq[k]

print(freq, f"\nMais repetidas ({m_freq}x):", [k for k in freq if freq[k] == m_freq])
```

**Questão 7: Boletim de uma turma**

```

turma = []
soma_g = 0
for _ in range(6):
    nome = input("Nome: ")
    med = (float(input("N1: ")) + float(input("N2: ")) + float(input("N3: "))) / 3
    turma.append({"nome": nome, "media": med})
    soma_g += med

m_geral = soma_g / 6
m_alta = -1
for a in turma:
    if a["media"] > m_alta: m_alta = a["media"]

print(turma, f"\nMédia Geral: {m_alta:.1f}")
print(">= Média Geral:", len([a for a in turma if a["media"] >= m_geral]))
print("Maior média:", [a["nome"] for a in turma if a["media"] == m_alta])
```

**Questão 8: Controle de estoque com movimentações**

```
est = [{"codigo": int(input("Cód: ")), "nome": input("Nome: "), "preco": float(input("Preço: ")), "quantidade": int(input("Qtd: "))} for _ in range(5)]

for _ in range(5):
    c, t, q = int(input("Cód: ")), input("Tipo (E/S): ").upper(), int(input("Qtd: "))
    for p in est:
        if p["codigo"] == c:
            if t == "E" and q > 0: p["quantidade"] += q
            elif t == "S" and 0 < q <= p["quantidade"]: p["quantidade"] -= q

tot = sum(p["preco"] * p["quantidade"] for p in est)
print("Estoque:", est, f"\nTotal: R$ {tot:.2f}")
```

**Questão 9: Ranking de jogadores**

```
jogs = []
m_tot, m_dif = -1, 999999

for _ in range(5):
    n = input("Nome: ")
    p = [float(input("P1: ")), float(input("P2: ")), float(input("P3: "))]
    tot = sum(p)

    maior_p, menor_p = p[0], p[0]
    for x in p:
        if x > maior_p: maior_p = x
        if x < menor_p: menor_p = x
    dif = maior_p - menor_p

    if tot > m_tot: m_tot = tot
    if dif < m_dif: m_dif = dif
    jogs.append({"nome": n, "total": tot, "dif": dif})

print("Campeão(es):", [j["nome"] for j in jogs if j["total"] == m_tot])
print("Mais regular(es):", [j["nome"] for j in jogs if j["dif"] == m_dif])
```

**Questão 10: Agenda de contatos sem duplicidade**

```
agenda = []
while (n := input("Nome ('FIM' sai): ")).upper() != "FIM":
    if any(c["nome"].lower() == n.lower() for c in agenda):
        print("Já existe!")
    else:
        agenda.append({"nome": n, "telefone": input("Tel: ")})

while (b := input("Buscar ('FIM' sai): ")).upper() != "FIM":
    res = [c["telefone"] for c in agenda if c["nome"].lower() == b.lower()]
    print(f"Tel: {res[0]}" if res else "Não encontrado.")
```

**Questão 11: Pesquisa de votos**

```
votos = {1: 0, 2: 0, 3: 0, 4: 0}
tot = 0

while (v := int(input("Voto 1-4 (0 sai): "))) != 0:
    if v in votos:
        votos[v] += 1
        tot += 1

m_votos = 0
for k in votos:
    if votos[k] > m_votos: m_votos = votos[k]

for k, v in votos.items():
    print(f"Opção {k}: {v} votos ({(v/tot*100 if tot else 0):.1f}%)")
print("Vencedor(es):", [k for k in votos if votos[k] == m_votos and tot > 0])
```

**Questão 12: Histórico de movimentações**

```
est = {101: 50, 102: 30, 103: 20}
hist = []

for _ in range(8):
    c, t, q = int(input("Cód: ")), input("Tipo (E/S): ").upper(), int(input("Qtd: "))
    ok = c in est and q > 0 and ((t == "E") or (t == "S" and est[c] >= q))
    if ok:
        est[c] += q if t == "E" else -q
    hist.append((c, t, q, "ACEITA" if ok else "RECUSADA"))

print("Estoque final:", est, "\nHistórico:", hist)
```

**Questão 13: Análise de vendas por vendedor**

```
eq = []
tot_g = 0

for _ in range(5):
    n = input("Nome: ")
    v = [float(input(f"Venda {i+1}: ")) for i in range(4)]
    t = sum(v)
    tot_g += t
    eq.append({"nome": n, "total": t})

med = tot_g / 5
m_tot = 0
for x in eq:
    if x["total"] > m_tot: m_tot = x["total"]

print(f"Total Geral: R$ {tot_g:.2f}")
print("Top vendedor(es):", [x["nome"] for x in eq if x["total"] == m_tot])
print("Acima da média:", len([x for x in eq if x["total"] > med]))
```

**Questão 14: Catálogo e pedidos**

```
prods = [{"codigo": int(input("Cód: ")), "nome": input("Nome: "), "preco": float(input("Preço: ")), "estoque": int(input("Qtd: "))} for _ in range(5)]
peds = [(int(input("Cód pedido: ")), int(input("Qtd: "))) for _ in range(6)]

tot = 0
for c, q in peds:
    atendido = False
    for p in prods:
        if p["codigo"] == c and p["estoque"] >= q:
            p["estoque"] -= q
            tot += q * p["preco"]
            atendido = True
            break
    print(f"Pedido ({c}, {q}):", "ATENDIDO" if atendido else "RECUSADO")

print(f"Total atendido: R$ {tot:.2f}")
```

**Questão 15: Sistema de vendas e relatório final**

```
prods = [{"codigo": int(input("Cód: ")), "nome": input("Nome: "), "preco": float(input("Preço: ")), "estoque": int(input("Qtd: ")), "vendidos": 0, "val_vendido": 0.0} for _ in range(5)]
hist = []

while (c := int(input("Cód venda (0 sai): "))) != 0:
    q = int(input("Qtd: "))
    p = next((x for x in prods if x["codigo"] == c), None)
    ok = p is not None and 0 < q <= p["estoque"]
    if ok:
        p["estoque"] -= q
        p["vendidos"] += q
        p["val_vendido"] += q * p["preco"]
        hist.append((c, q, q * p["preco"], "ACEITA"))
    else:
        hist.append((c, q, 0.0, "RECUSADA"))

m_val = max(x["val_vendido"] for x in prods)
tot_v = sum(1 for h in hist if h[3] == "ACEITA")
tot_u = sum(x["vendidos"] for x in prods)
tot_r = sum(x["val_vendido"] for x in prods)

print("Mais vendidos:", [x["nome"] for x in prods if x["val_vendido"] == m_val > 0])
print("Sem vendas:", [x["nome"] for x in prods if x["vendidos"] == 0])
print("Resumo (Vendas, Unidades, Total R$):", (tot_v, tot_u, tot_r))

# Dicionários: mapeiam atributos do produto.
# Tuplas: guardam registros fixos/imutáveis (histórico e resumo).
# Listas: armazenam coleções dinâmicas de dados.
```
