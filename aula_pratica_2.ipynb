questão 2. Estatísticas de uma turma

```python
soma = 0
aprovados = 0
abaixo = 0
for i in range(5):
   while True:
    try:
         nota = float(input(f"nota do aluno{i+1}:"))
         if not 0 <= nota <= 10:
            print("nota deve estar entre 0 e 10.")
            continue
         break
    except ValueError:
      print("digite uma nota numerica valida.")

soma += nota
if nota >= 7:
  aprovados += 1
else:
  abaixo += 1
  print("madia: ",soma / 5)
  print("notas >= 7:",aprovados)
  print("notas < 7:",abaixo)
```

questão 5.Maquina de venda de ingresso

```python
preço = 25
total = 0

while total < preço:
    try:
         deposito = int(input("insira um valor inteiro: "))
         if deposito > 0:
             total += deposito
         else:
          print("o valor deve ser positivo.")
    except ValueError:
         print("insira um valor inteiro valido.")

print(f"Total inserido: R${total:.2f}")
print(f"Troco: R${total - preço:.2f}")
```

questão 7.Maior e segundo maior

```python
vistos = set()
maior = None
segundo_maior = None

for i in range(8):
    while True:
        try:
            numero = int(input(f"Digite o {i + 1}º número: "))
            if numero in vistos:
                print("Esse número já foi digitado. Digite outro.")
                continue
            vistos.add(numero)
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if maior is None or numero > maior:
        segundo_maior = maior
        maior = numero
    elif segundo_maior is None or numero > segundo_maior:
        segundo_maior = numero

print("Maior:", maior)
print("Segundo maior:", segundo_maior)
```

questão 8.Repetição consecutiva

```python
anterior = None
tem_repeticao = False

for i in range(8):
    while True:
        try:
            numero = int(input(f"Digite o {i + 1}º número: "))
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if anterior is not None and numero == anterior:
        tem_repeticao = True

    anterior = numero

if tem_repeticao:
    print("Existem valores iguais consecutivos.")
else:
    print("Não existem valores iguais consecutivos.")
```

questão 9.Controle de saldo

```python
saldo = 500.0

for i in range(6):
    while True:
        try:
            tipo = int(input(f"Operação {i + 1} (1 = depósito; 2 = saque): "))
            if tipo == 1 or tipo == 2:
                break
            print("Digite 1 para depósito ou 2 para saque.")
        except ValueError:
            print("Digite um código inteiro válido.")

    while True:
        try:
            valor = float(input("Digite o valor: R$ "))
            if valor > 0:
                break
            print("O valor deve ser positivo.")
        except ValueError:
            print("Digite um valor numérico válido.")

    if tipo == 1:
        saldo += valor
    elif valor <= saldo:
        saldo -= valor
    else:
        print("Saldo insuficiente.")

print(f"Saldo final: R$ {saldo:.2f}")
```

questão 10.Menor quantidade de cedulas

```python
while True:
    try:
        valor = int(input("Digite um valor inteiro positivo: "))
        if valor > 0:
            break
        print("O valor deve ser positivo.")
    except ValueError:
        print("Digite um número inteiro válido.")

restante = valor

for cedula in (100, 50, 20, 10, 5, 2, 1):
    quantidade = restante // cedula
    restante = restante % cedula
    print(f"Cédulas de R$ {cedula}: {quantidade}")
```

questão 11.numero primo

```python
while True:
    try:
        numero = int(input("Digite um inteiro maior que 1: "))
        if numero > 1:
            break
        print("O número deve ser maior que 1.")
    except ValueError:
        print("Digite um número inteiro válido.")

primo = True
divisor = 2

while divisor <= numero // divisor:
    if numero % divisor == 0:
        primo = False
        break
    divisor += 1

if primo:
    print("É primo.")
else:
    print("Não é primo.")
```

questão 12.Divisores de um numero

```python
while True:
    try:
        n = int(input("Digite um inteiro positivo: "))
        if n > 0:
            break
        print("O número deve ser positivo.")
    except ValueError:
        print("Digite um número inteiro válido.")

quantidade = 0

print("Divisores positivos:")
for divisor in range(1, n + 1):
    if n % divisor == 0:
        print(divisor)
        quantidade += 1

print("Quantidade de divisores:", quantidade)
```

questão 13.Primos no intervalo

```python
while True:
    try:
        a = int(input("Digite A, positivo: "))
        if a > 0:
            break
        print("A deve ser positivo.")
    except ValueError:
        print("Digite um inteiro válido.")

while True:
    try:
        b = int(input("Digite B, maior que A: "))
        if b > a:
            break
        print("B deve ser maior que A.")
    except ValueError:
        print("Digite um inteiro válido.")

quantidade = 0

print("Primos encontrados:")
for numero in range(a, b + 1):
    if numero > 1:
        primo = True
        divisor = 2

        while divisor <= numero // divisor:
            if numero % divisor == 0:
                primo = False
                break
            divisor += 1

        if primo:
            print(numero)
            quantidade += 1

print("Quantidade de primos:", quantidade)
```

questão 14.Jogo com limite de alternativa

```python
secreto = 37
tentativas = 0
acertou = False

while tentativas < 5 and not acertou:
    try:
        palpite = int(input("Digite seu palpite: "))
    except ValueError:
        print("Digite um número inteiro válido.")
        continue

    tentativas += 1

    if palpite == secreto:
        print("Você acertou!")
        acertou = True
    elif palpite < secreto:
        print("O número secreto é maior.")
    else:
        print("O número secreto é menor.")

if not acertou:
    print("Suas tentativas acabaram. O número era 37.")

```

questão 15.Caixa de atendimento

```python
saldo = 200.0
quantidade_entradas = 0
retiradas_realizadas = 0
retiradas_recusadas = 0
total_entrou = 0.0
total_saiu = 0.0

while True:
    while True:
        try:
            codigo = int(input("Código (1 = entrada; 2 = retirada; 0 = encerrar): "))
            if codigo in (0, 1, 2):
                break
            print("Código inválido. Digite 0, 1 ou 2.")
        except ValueError:
            print("Digite um código inteiro válido.")

    if codigo == 0:
        break

    while True:
        try:
            valor = float(input("Valor: R$ "))
            if valor > 0:
                break
            print("O valor deve ser positivo.")
        except ValueError:
            print("Digite um valor numérico válido.")

    if codigo == 1:
        saldo += valor
        total_entrou += valor
        quantidade_entradas += 1
    elif valor <= saldo:
        saldo -= valor
        total_saiu += valor
        retiradas_realizadas += 1
    else:
        print("Retirada recusada: saldo insuficiente.")
        retiradas_recusadas += 1

print(f"Saldo final: R$ {saldo:.2f}")
print("Quantidade de entradas:", quantidade_entradas)
print("Retiradas realizadas:", retiradas_realizadas)
print("Retiradas recusadas:", retiradas_recusadas)
print(f"Total que entrou: R$ {total_entrou:.2f}")
print(f"Total que saiu: R$ {total_saiu:.2f}")
```
