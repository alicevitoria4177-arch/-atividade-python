```python
numeros = []
for i in range(10):
    num = int(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

pares = []
impares = []

for num in numeros:
    if num % 2 == 0:
        pares.append(num)
    else:
        impares.append(num)

print(f"Lista completa: {numeros}")
print(f"Pares: {pares}")
print(f"Ímpares: {impares}")
```

```python
numeros = []
for i in range(8):
    num = int(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

maior = numeros[0]
menor = numeros[0]
pos_maior = 0
pos_menor = 0

for i in range(1, len(numeros)):
    if numeros[i] > maior:
        maior = numeros[i]
        pos_maior = i
    if numeros[i] < menor:
        menor = numeros[i]
        pos_menor = i

print(f"Maior valor: {maior} (primeira ocorrência no índice {pos_maior})")
print(f"Menor valor: {menor} (primeira ocorrência no índice {pos_menor})")
```

```python
valores = []
for i in range(6):
    num = int(input(f"Digite o {i+1}º número: "))
    valores.append(num)

tupla = tuple(valores)

qtd_cinco = tupla.count(5)
pares = 0
for x in tupla:
    if x % 2 == 0:
        pares += 1

print(f"Tupla: {tupla}")
print(f"O número 5 aparece {qtd_cinco} vez(es).")

if 5 in tupla:
    print(f"A primeira ocorrência do 5 está na posição (índice) {tupla.index(5)}.")
else:
    print("O número 5 não foi digitado.")

print(f"Quantidade de números pares: {pares}")
```

```python
aluno = {}
aluno['nome'] = input("Digite o nome: ")
aluno['turma'] = input("Digite a turma: ")
aluno['nota1'] = float(input("Digite a 1ª nota: "))
aluno['nota2'] = float(input("Digite a 2ª nota: "))

aluno['media'] = (aluno['nota1'] + aluno['nota2']) / 2

if aluno['media'] >= 7:
    aluno['situacao'] = 'Aprovado'
elif aluno['media'] >= 5:
    aluno['situacao'] = 'Recuperação'
else:
    aluno['situacao'] = 'Reprovado'

print("\n--- Dados do Aluno ---")
for chave, valor in aluno.items():
    print(f"{chave.capitalize()}: {valor}")
```

QUESTÃO 12

```python
estoque = {
    "caderno": 10,
    "caneta": 50,
    "borracha": 30,
    "mochila": 5
}

prod = input("Digite o nome do produto vendido: ").strip().lower()
qtd_vendida = int(input("Digite a quantidade vendida: "))

if prod in estoque:
    if estoque[prod] >= qtd_vendida:
        estoque[prod] -= qtd_vendida
        print("Venda realizada com sucesso!")
    else:
        print("Erro: Estoque insuficiente.")
else:
    print("Erro: Produto não encontrado no estoque.")

print(f"Estoque final: {estoque}")
```

```python
turma = []

for i in range(5):
    print(f"\n--- Cadastro do Aluno {i+1} ---")
    nome = input("Nome: ")
    n1 = float(input("Nota 1: "))
    n2 = float(input("Nota 2: "))
    media = (n1 + n2) / 2

    aluno = {
        "nome": nome,
        "nota1": n1,
        "nota2": n2,
        "media": media
    }
    turma.append(aluno)

print("\n--- Lista de Todos os Alunos ---")
for a in turma:
    print(f"Aluno: {a['nome']} | Média: {a['media']:.2f}")
```

QUESTÃO 14

```python
catalogo = []

print("--- Cadastro de 5 Produtos ---")
for i in range(5):
    codigo = int(input(f"Código do produto {i+1}: "))
    nome = input(f"Nome do produto {i+1}: ")
    preco = float(input(f"Preço do produto {i+1}: "))
    produto = (codigo, nome, preco)
    catalogo.append(produto)

cod_busca = int(input("\nDigite o código do produto para buscar: "))
encontrado = False

for prod in catalogo:
    if prod[0] == cod_busca:
        print(f"Produto encontrado: {prod[1]} | Preço: R$ {prod[2]:.2f}")
        encontrado = True
        break

if not encontrado:
    print("Código de produto não encontrado.")
```

QUESTÃO 15

```python
estoque = []

print("--- Cadastro de 5 Itens no Estoque ---")
for i in range(5):
    print(f"\nItem {i+1}:")
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    quantidade = int(input("Quantidade: "))
    preco = float(input("Preço unitário: "))

    item = {
        "codigo": codigo,
        "nome": nome,
        "quantidade": quantidade,
        "preco": preco
    }
    estoque.append(item)

# (a) Mostrar todos os itens
print("\n--- (a) Itens Cadastrados ---")
for item in estoque:
    print(f"Código: {item['codigo']} | Nome: {item['nome']} | Qtd: {item['quantidade']} | Preço: R$ {item['preco']:.2f}")

# (b) Calcular o valor total do estoque e (c) identificar o item de maior valor
valor_total_estoque = 0.0
item_maior_valor = None
maior_valor = -1.0

for item in estoque:
    valor_item = item['quantidade'] * item['preco']
    valor_total_estoque += valor_item

    if valor_item > maior_valor:
        maior_valor = valor_item
        item_maior_valor = item

print(f"\n--- (b) Valor Total do Estoque ---")
print(f"R$ {valor_total_estoque:.2f}")

print(f"\n--- (c) Item de Maior Valor em Estoque (Qtd x Preço) ---")
print(f"Produto: {item_maior_valor['nome']} | Valor Total: R$ {maior_valor:.2f}")

# (d) Criar tupla resumo
tupla_resumo = (len(estoque), valor_total_estoque)

print(f"\n--- (d) Tupla Resumo ---")
print(f"(Qtd de itens cadastrados, Valor total do estoque): {tupla_resumo}")
```
