QUESTÃO 1

```python
primeiro = int(input("Digite um número: "))

maior = primeiro
menor = primeiro

for i in range(5):
    numero = int(input("Digite um número: "))

    if numero > maior:
        maior = numero

    if numero < menor:
        menor = numero

print("Maior:", maior)
print("Menor:", menor)
```

QUESTÃO 2

```python
soma = 0
aprovados = 0
abaixo = 0

for i in range(5):
    nota = float(input("Nota: "))

    soma += nota

    if nota >= 7:
        aprovados += 1
    else:
        abaixo += 1

media = soma / 5

print("Média:", media)
print("Notas >= 7:", aprovados)
print("Notas < 7:", abaixo)
```

```python
maior = None
segundo_maior = None

for i in range(8):
    numero = int(input("Digite um número: "))

    if maior is None:
        maior = numero

    elif numero > maior:
        segundo_maior = maior
        maior = numero

    elif segundo_maior is None or numero > segundo_maior:
        segundo_maior = numero

print("Maior:", maior)
print("Segundo maior:", segundo_maior)
```

```python
anterior = int(input("Digite um número: "))
repetiu = False

for i in range(7):
    numero = int(input("Digite um número: "))

    if numero == anterior:
        repetiu = True

    anterior = numero

if repetiu:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

QUESTÃO 9

```python
saldo = 500

for i in range(6):
    tipo = int(input("Digite o tipo (1 = depósito, 2 = saque): "))
    valor = float(input("Digite o valor: "))

    if tipo == 1:
        saldo += valor

    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente.")

    else:
        print("Tipo de operação inválido.")

print("Saldo final: R$", saldo)
```

QUESTÃO 13

```python
A = int(input("Digite A: "))
B = int(input("Digite B: "))

quantidade = 0

for numero in range(A, B + 1):

    if numero > 1:
        primo = True

        for divisor in range(2, numero):
            if numero % divisor == 0:
                primo = False
                break

        if primo:
            print(numero)
            quantidade += 1

print("Quantidade de primos:", quantidade)
```

QUESTÃO 14

```python
secreto = 37
acertou = False

for tentativa in range(1, 6):
    numero = int(input("Tentativa " + str(tentativa) + ": "))

    if numero == secreto:
        print("Parabéns! Você acertou!")
        acertou = True
        break

    elif numero < secreto:
        print("O segredo é maior.")

    else:
        print("O segredo é menor.")

if not acertou:
    print("Você perdeu! O número era", secreto)
```

QUESTÃO 15

```python
saldo = 200

quantidade_entradas = 0
quantidade_retiradas = 0
quantidade_recusadas = 0

total_entradas = 0
total_saidas = 0

while True:
    codigo = int(input("Digite o código (0 = sair, 1 = entrada, 2 = retirada): "))

    if codigo == 0:
        break

    elif codigo == 1:
        valor = float(input("Digite o valor da entrada: "))

        saldo += valor
        quantidade_entradas += 1
        total_entradas += valor

    elif codigo == 2:
        valor = float(input("Digite o valor da retirada: "))

        if valor <= saldo:
            saldo -= valor
            quantidade_retiradas += 1
            total_saidas += valor
        else:
            print("Retirada recusada: saldo insuficiente.")
            quantidade_recusadas += 1

    else:
        print("Código inválido.")

print("\n--- RESUMO ---")
print("Saldo final: R$", saldo)
print("Quantidade de entradas:", quantidade_entradas)
print("Quantidade de retiradas realizadas:", quantidade_retiradas)
print("Quantidade de retiradas recusadas:", quantidade_recusadas)
print("Total que entrou: R$", total_entradas)
print("Total que saiu: R$", total_saidas)
```
