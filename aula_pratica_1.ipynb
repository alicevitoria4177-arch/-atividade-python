questão 1 . positivo , negativo ou zero.

```python
while True:
  try:
    numero = int(input("digite um numero: "))
    break
  except ValueError:
    print("digite um valor valido.")

if numero > 0:
  print("positivo")
elif numero < 0:
  print("negativo")
else:
  print("zero")
```

questão 2 . maior entre dois numeros.

```python
valores = []
for i in range(2):
    while True:
      try:
          n = int(input(f"digite o {i+1}º numero: "))
          valores.append(n)
          break
      except ValueError:
        print("informe um numero valido")

a, b = valores
if a > b:
  print("maior: ",a)
elif b > a:
  print("maior: ",b)
else:
  print("os numeros sao iguais")
```

questão 5. situação do aluno

```python
notas = []
for i in range(2):
    while True:
        try:
          nota = float(input(f"digite a {i+1}º nota: "))
          if 0 <= nota <= 10:
            notas.append(nota)
            break
          print("a nota deve estar entre 0 e 10")
        except ValueError:
          print("digite uma nota numerica valida")

media = sum(notas) / 2
if media >= 7:
  print("aprovado")
elif media >= 5:
  print("recuperação")
else:
  print("reprovado")
print("media: ",media)
```

questão 9. contar quantos dos 10 são pares

```python
pares = 0

for i in range(10):
    while True:
        try:
            numero = int(input(f"Digite o {i + 1}º número: "))
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if numero % 2 == 0:
        pares += 1

print("Quantidade de pares:", pares)
```

Questão 11.Maior de 6 números

```python
maior = None

for i in range(6):
    while True:
        try:
            numero = int(input(f"Digite o {i + 1}º número: "))
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if maior is None or numero > maior:
        maior = numero

print("Maior número:", maior)
```

Questão 14.Adivinhar o número secreto

```python
secreto = 7
tentativas = 0

while True:
    try:
        chute = int(input("Adivinhe o número secreto: "))
        tentativas += 1
    except ValueError:
        print("Digite um número inteiro válido.")
        continue

    if chute == secreto:
        print(f"Acertou em {tentativas} tentativa(s)!")
        break
    elif chute < secreto:
        print("O número secreto é maior.")
    else:
        print("O número secreto é menor.")
```

Questão 15. Analisar números até digitar 0

```python

quantidade = 0
positivos = 0
negativos = 0
soma = 0

while True:
    try:
        numero = int(input("Digite um número inteiro (0 para parar): "))
    except ValueError:
        print("Digite um número inteiro válido.")
        continue

    if numero == 0:
        break

    quantidade += 1
    soma += numero

    if numero > 0:
        positivos += 1
    else:
        negativos += 1

print("Quantidade:", quantidade)
print("Positivos:", positivos)
print("Negativos:", negativos)
print("Soma:", soma)
```
